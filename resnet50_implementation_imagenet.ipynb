```python
# import necessary libraries
from keras.applications.resnet50 import ResNet50
from keras.applications.resnet50 import preprocess_input, decode_predictions
from tensorflow.keras.utils import load_img, img_to_array
import numpy as np
```

```python
# Load pre trained model
model = ResNet50(weights='imagenet')
# weights= 'imagenet' ensures we get imagenet trained weights
```

```python
# Load and preprocess image
# img = load_img('/kaggle/input/datasets/naveenkumartiwari/cat-image/cute-ginger-cat-sitting-looking-camera-isolated-white-background_270100-2630.avif', target_size = (224, 224))
img = load_img('/kaggle/input/datasets/naveenkumartiwari/cat-image/cute-ginger-cat-sitting-looking-camera-isolated-white-background_270100-2630.avif', target_size = (224, 224))
x = img_to_array(img)
x = np.expand_dims(x, axis = 0)
x = preprocess_input(x)
```

```python
# MAKE PREDICTION
predictions = model.predict(x)
```

```python
decoded  = decode_predictions(predictions, top = 1)
```
